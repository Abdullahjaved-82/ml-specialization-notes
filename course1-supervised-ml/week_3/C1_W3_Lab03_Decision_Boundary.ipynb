# Logistic Regression: Decision Boundary

## Goals
Plot the decision boundary for a logistic regression model to see what it's actually predicting.

```python
import numpy as np
import matplotlib.pyplot as plt

#sigmoid fucntion
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

## Dataset
X has 6 examples, each with two features. y is 0 or 1.

```python
X = np.array([[0.5,1.5],[1,1],[1.5,0.5],[3,0.5],[2,2],[1,2.5]])
y = np.array([0,0,0,1,1,1]).reshape(-1,1)  #.reshape for converting into matrix , as rows are -1 mean not defined , so they will be decided acording to coulumn
```

### Plot data
Points with y=1 shown as red X's, y=0 as blue circles.

```python
fig, ax = plt.subplots(1, 1, figsize=(4, 4))

y_flat = y.flatten()
pos = y_flat == 1
neg = y_flat == 0

ax.scatter(X[pos, 0], X[pos, 1], marker='x', s=80, c='red', label='y=1')
ax.scatter(X[neg, 0], X[neg, 1], marker='o', s=100, facecolors='none', edgecolors='blue', label='y=0')

ax.axis([0, 4, 0, 3.5])
ax.set_ylabel('$x_1$')
ax.set_xlabel('$x_0$')
ax.legend()
plt.savefig("decision_boundary_data.png")
plt.close()

from IPython.display import Image
Image("decision_boundary_data.png")
```

## Logistic Regression Model

Model: f(x) = g(w0*x0 + w1*x1 + b), where g is sigmoid.

Say we trained the model and got b=-3, w0=1, w1=1. So f(x) = g(x0 + x1 - 3).

Recall: g(z) >= 0.5 when z >= 0. So the model predicts y=1 when x0+x1-3 >= 0, and y=0 otherwise.

```python
z = np.arange(-10, 11)

fig, ax = plt.subplots(1, 1, figsize=(5, 3))
ax.plot(z, sigmoid(z), c="b")
ax.set_title("Sigmoid function")
ax.set_ylabel('sigmoid(z)')
ax.set_xlabel('z')
ax.axhline(0.5, color='gray', linestyle='--')
plt.savefig("sigmoid_recheck.png")
plt.close()

from IPython.display import Image
Image("sigmoid_recheck.png")
```

### Plotting decision boundary

Our model is f(x) = g(x0 + x1 - 3). It predicts y=1 when x0+x1-3 >= 0.

The boundary line itself is where x0+x1-3 = 0, which rearranges to x1 = 3 - x0.

```python
x0 = np.arange(0, 6)
x1 = 3 - x0      #applying boundry formulla

fig, ax = plt.subplots(1, 1, figsize=(5, 4))
ax.plot(x0, x1, c="b")
ax.axis([0, 4, 0, 3.5])  #defining graph range
ax.fill_between(x0, x1, alpha=0.2)

y_flat = y.flatten()
pos = y_flat == 1
neg = y_flat == 0
ax.scatter(X[pos, 0], X[pos, 1], marker='x', s=80, c='red', label='y=1')
ax.scatter(X[neg, 0], X[neg, 1], marker='o', s=100, facecolors='none', edgecolors='blue', label='y=0')

ax.set_ylabel('$x_1$')
ax.set_xlabel('$x_0$')
plt.savefig("decision_boundary.png")
plt.close()

from IPython.display import Image
Image("decision_boundary.png")
```
