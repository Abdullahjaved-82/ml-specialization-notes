# Logistic Regression: The Sigmoid Function

## Goals
Explore the sigmoid function, and see how it fixes linear regression's failure to handle classification.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Sigmoid Function

Linear regression's prediction (w*x + b) can be any number - huge negative to huge positive. For classification we want output confined between 0 and 1. The sigmoid function does exactly that.

Formula: g(z) = 1 / (1 + e^(-z))

NumPy's exp() calculates e raised to a power, for every element in an array (or a single number).

```python
input_array = np.array([1,2,3])
exp_array = np.exp([input_array]) #e is a special mathematical constant, roughly 2.71828, this is used for conversion
print("Input to exp:",input_array)
print("Output of exp:", exp_array)

input_val = 1
exp_val = np.exp(input_val)
print("Input to exp:", input_val)
print("Output of exp:", exp_val)
```

### The sigmoid function
Implementing g(z) = 1 / (1 + e^(-z))

```python
def sigmoid(z):
    """
    Compute the sigmoid of z

    Args:
        z (ndarray): A scalar, numpy array of any size.

    Returns:
        g (ndarray): sigmoid(z), with the same shape as z
    """
    g = 1 / (1 + np.exp(-z))
    return g
```

Let's see the output for a range of z values, from -10 to 10.

```python
z_tmp = np.arange(-10,11)
y = sigmoid(z_tmp)

np.set_printoptions(precision=3)
print("Input (z), Output (sigmoid(z))")
print(np.c_[z_tmp, y])   #.c_ for stacking two seprate array into two coulmns
```

Now let's plot this function.

```python
fig,ax = plt.subplots(1,1,figsize=(5,3))
ax.plot(z_tmp, y, c="b")
ax.set_title("Sigmoid function")
ax.set_ylabel('sigmoid(z)')
ax.set_xlabel('z')
ax.axhline(0.5, color='gray', linestyle='--')
plt.savefig("sigmoid_curve.png")
plt.close()

from IPython.display import Image
Image("sigmoid_curve.png")
```

## Summary
The sigmoid function maps any real number to a value between 0 and 1, making it suitable for classification. Logistic regression combines linear regression (z = w*x+b) with sigmoid to produce probability-like outputs, fixing the failure seen with plain linear regression.
