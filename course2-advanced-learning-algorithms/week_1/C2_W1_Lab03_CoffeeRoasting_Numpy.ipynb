# Coffee Roasting - NumPy Implementation
Building the same coffee roasting network from the previous lab, but implementing forward propagation by hand in NumPy instead of using TensorFlow/Keras.

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
def load_coffee_data():
    np.random.seed(2)
    X = np.random.rand(400,2)
    X[:,0] = X[:,0] * (285-150) + 150
    X[:,1] = X[:,1] * 4 + 11.5
    Y = np.zeros(len(X))
    for i in range(len(X)):
        t,d = X[i]
        y = -3/(260-175)*t + 21
        if (t > 175 and t < 260 and d > 12 and d < 15 and d <= y):
            Y[i] = 1
    return X, Y.reshape(-1,1)

def sigmoid(z):
    return 1/(1+np.exp(-z))

X, Y = load_coffee_data()
print(X.shape, Y.shape)

norm_l = tf.keras.layers.Normalization(axis=-1)
norm_l.adapt(X)
Xn = norm_l(X)
```

### Activation function
Define the sigmoid activation function to use inside the hand-built dense layer.

```python
g = sigmoid
```

```python
def my_dense(a_in, w, b):
    units = w.shape[1]
    a_out = np.zeros(units)
    for j in range(units):
        wj = w[:,j]
        z = np.dot(wj,a_in)+b[j]
        a_out[j] = g(z)
    return(a_out)
```

### my_sequential
Stack two my_dense layers together, passing the output of the first as input to the second — the manual equivalent of Keras's Sequential model.

```python
def my_sequential(x,W1,b1,W2,b2):
    a1 = my_dense(x,W1,b1)
    a2 = my_dense(a1,W2,b2)
    return(a2)
    
```
