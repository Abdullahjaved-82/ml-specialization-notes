# Backpropagation using a Computation Graph
Backprop breaks a complex derivative into small steps, computing local derivatives node-by-node and chaining them together (chain rule) to get the full derivative.

Example: J = (2+3w)^2, broken into a = 2+3w, then J = a^2.

```python
from sympy import symbols, diff

w = 3
a = 2 + 3*w
J = a**2
print(f"a = {a}, J = {J}")
```

```python
a_epsilon = a + 0.001
J_epsilon = a_epsilon**2
k = (J_epsilon - J)/0.001
print(f"dJ_da ~= k = {k}")

sw, sJ, sa = symbols('w,J,a')
sJ = sa**2
dJ_da = diff(sJ, sa)
print(dJ_da, dJ_da.subs([(sa,a)]))
```

### da/dw and the chain rule
a = 2+3w, so da/dw = 3. Chain rule: dJ/dw = da/dw × dJ/da.

```python
sa = 2 + 3*sw
da_dw = diff(sa, sw)
print("da_dw =", da_dw)

dJ_dw = da_dw * dJ_da
print("dJ_dw =", dJ_dw, "=", dJ_dw.subs([(sa,a)]))
```

### A larger computation graph
Network: c=wx, a=c+b, d=a-y, J=d²/2. Forward pass first.

```python
x = 2
w = -2
b = 8
y = 1

c = w * x
a = c + b
d = a - y
J = d**2/2
print(f"J={J}, d={d}, a={a}, c={c}")
```

### Backward propagation
Compute local derivatives at each node, right to left, chaining via multiplication.

```python
sx, sw, sb, sy, sJ, sa, sc, sd = symbols('x,w,b,y,J,a,c,d')

# J = d^2/2  ->  dJ/dd
sJ = sd**2/2
dJ_dd = diff(sJ, sd)

# d = a - y  ->  dd/da
sd = sa - sy
dd_da = diff(sd, sa)
dJ_da = dd_da * dJ_dd

# a = c + b  ->  da/dc, da/db
sa = sc + sb
da_dc = diff(sa, sc)
da_db = diff(sa, sb)
dJ_dc = da_dc * dJ_da
dJ_db = da_db * dJ_da

# c = w * x  ->  dc/dw
sc = sw * sx
dc_dw = diff(sc, sw)
dJ_dw = dc_dw * dJ_dc

print("dJ_dd =", dJ_dd)
print("dJ_da =", dJ_da)
print("dJ_dc =", dJ_dc, " dJ_db =", dJ_db)
print("dJ_dw =", dJ_dw.subs([(sd,d),(sx,x)]))
```

### Summary
This lab implemented backpropagation manually using SymPy, breaking a computation into small nodes (c, a, d, J) and computing each node's local derivative, then chaining them via the chain rule to find dJ/dw. This is the same algorithm TensorFlow runs automatically during model.fit() to compute gradients for every parameter in a network.
