# Derivatives
A derivative describes how much a function's output changes when its input changes by a small amount. Below: computing derivatives two ways — arithmetically (small epsilon) and symbolically (SymPy).

```python
from sympy import symbols,diff

J = (3)**2
J_epsilon = (3 + 0.001)**2
k = (J_epsilon - J)/0.001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k:0.6f}")
```

```python
J, w = symbols('J, w')
J = w**2
dJ_dw = diff(J, w)
print(dJ_dw)
print(dJ_dw.subs([(w,3)]))
```

### Summary
Derivatives were computed two ways: arithmetically using a small epsilon, and symbolically using SymPy's diff() function. Both approaches confirm that a derivative tells you the rate of change of a function's output relative to its input — the core mechanism behind gradient descent's weight updates.
